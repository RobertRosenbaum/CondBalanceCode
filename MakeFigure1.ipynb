```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

EffClr = [0.2,0.2,0.8]
CurrClr = [0.8,0.2,0.2]
CondClr = [0,0,0]
CorrColor = [.15,.6,.15]
AsynchColor = [.6,.2,.6]

XClr = [.2,.75,.2]

VClr = [.4,.4,.4]

plt.rcParams['svg.fonttype'] = 'none'
plt.rcParams['font.size']=8

DataDir = './Data/'
ImagesDir = './Image/'

VsNFilename = 'VsTData.npy'
```

```python
plotdata = np.load(DataDir+VsNFilename, allow_pickle=True).item()

T=plotdata['T']
dt=plotdata['dt']
Nt=plotdata['Nt']
Nburn=plotdata['Nburn']
Tburn=plotdata['Tburn']
SxAsynch=plotdata['SxAsynch']
SxCorr=plotdata['SxCorr']
Ve=plotdata['Ve']
dtRecord=plotdata['dtRecord']

time=np.arange(0,T,dt)
```

```python

fig, axes = plt.subplot_mosaic("aacc;aacc;bbdd;bbdd",
  figsize=(3.4,3.4))#figsize=(1.05*4.8,1.05*3.5))


c0='a'
ax0 = axes[c0]
# Remove axis clipping so that the image can spill over the subplot axes.
im = plt.imread(ImagesDir+'NetworkDiagram.png')
im_artist = ax0.imshow(im, aspect=1, extent=[-1.5, 1.5, -1.5, 1.5], zorder=10, clip_on=False)
for spine in ax0.spines.values():
    spine.set_visible(False)
ax0.set_clip_on(False)
ax0.set_xlim(-1.2, 1.2)  
ax0.set_ylim(-1.2, 1.2)
ax0.axis('off')
ax0.set_title('('+c0+')',loc='left')


c0='b'
ax0 = axes[c0]
ax0.plot(time[Nburn:]-Tburn,Ve[Nburn:], color=VClr)
# Draw a scale bar "corner" with lengths 10 (x) and 5 (y)
scalex = 100
scaley = 25
# Place the corner at the lower right corner inside the plot
x0 = (time[Nburn:]-Tburn).max() - scalex - 0  # 5 ms inset
y0 = Ve[Nburn:].min() + 40  # 5 mV inset from bottom
# Horizontal scale bar
ax0.plot([x0, x0 + scalex], [y0, y0], color=VClr, lw=1.5, clip_on=False)
# Vertical scale bar
ax0.plot([x0, x0], [y0, y0 + scaley], color=VClr, lw=1.5, clip_on=False)
# Add scale labels
ax0.text(x0 + scalex/2, y0-3.5, f'{scalex} ms', ha='center', va='top', fontsize=8)
ax0.text(x0-3.5, y0 + scaley/2+3, f'{scaley} mV', ha='right', va='center', rotation='vertical', fontsize=8)
ax0.axis('off')
sns.despine(ax=ax0)
ax0.set_title('('+c0+')',loc='left')

c0='c'
ax0 = axes[c0]
numSPlot = 200
iPlot = np.where(SxAsynch[1,:]<numSPlot)[0]
ax0.plot(SxAsynch[0,iPlot],SxAsynch[1,iPlot],'.',color=VClr,markersize=1)
ax0.set_xlabel('')
ax0.set_ylabel('x neuron index')
ax0.set_xticks([0,500])
ax0.set_title('c=0')
ax0.set_xlim(0,T)
ax0.set_ylim(0,numSPlot)
#ax0.axis('off')
#sns.despine(ax=ax0)
ax0.set_title('('+c0+')',loc='left')


c0='d'
ax0 = axes[c0]
numSPlot = 200
iPlot = np.where(SxCorr[1,:]<numSPlot)[0]
ax0.plot(SxCorr[0,iPlot],SxCorr[1,iPlot],'.',color=VClr,markersize=1)
ax0.set_xlabel('time (ms)', labelpad=-8)
ax0.set_ylabel('x neuron index')
ax0.set_xticks([0,500])
ax0.set_title('c>0')
ax0.set_xlim(0,T)
ax0.set_ylim(0,numSPlot)
#ax0.axis('off')
#sns.despine(ax=ax0)
ax0.set_title('('+c0+')',loc='left')


plt.tight_layout()


plt.savefig(ImagesDir+'Figure1.pdf')
```
